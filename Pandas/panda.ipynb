```python
# QUICK EXAMPLE OF READING A Anime CSV FILE
data3 = pd.read_csv(r'pandas_anime.csv')
print(data3)
```

```python
data3.head(8)
```

```python
# make a new column for episode count
def episode_count(txt):
    episodes =''
    check = False
    for i in range(len(txt)):
        if txt[i] == ')':
            break
        if txt[i] == '(':
            check = True
            continue
        if check == True:
            episodes += txt[i]
    return episodes
data3['episode'] = data3['Title'].apply(episode_count)   
data3['episode'] = data3['episode'].str.replace("eps","")
data3['episode'] = data3['episode'].astype('int')
```

```python
data3
```

```python
# make a new column for time stamp
def time_stamp(txt):
    time =''
    check = False
    for i in range(len(txt)):
        if txt[i] == ')':
            for j in range(i+1,i+20):
                time += txt[j]
            break
    return time
data3['title_time'] = data3['Title'].apply(time_stamp)
data3
```

```python
# which anime has highest score
data3[data3['Score'] == data3['Score'].max()]
```

```python
# give me top 5 anime based on score
data3['Title'].head()
```

```python
# which anime has highest episode count
data3[data3['episode'] == data3['episode'].max()]
```

```python
# QUICK EXAMPLE OF READING A Countries CSV FILE
dat = pd.read_csv(r'pandas_Countries.csv')
data =pd.DataFrame(dat)
data
```

```python
# which countries has highest population
data[data['population'] == data['population'].max()]
```

```python
# what is the capital of the country with highset popuation
data[ data['population']==data['population'].max()]['capital_city']
```

```python
# which country has least population
data[data['population'] == data['population'].min()]
```

```python
# what is the capital of the country with least population
data[ data['population'] == data['population'].min()]['capital_city']
```

```python
# give me top 5 countries with highest democratic score
top5 = data.sort_values(by='democracy_score', ascending=False).head(5)
top5
```

```python
# how many regions are there
data['region'].value_counts().count()
```

```python
#how many countries lie in Eastern Europe region
data[data['region'] == 'Eastern Europe']['country'].count()
```

```python
#who is the political leader of the 2nd highest populated country
data.nlargest(2,'population').iloc[1]['political_leader']
```

```python
# how many countries are there whose political leaders are unknown
data[data['political_leader'].isnull()].shape[0]
```

```python
#how many country have Republic in their full name
count = 0
for i in range(len(data)):
    if 'Republic'.lower() in data.iloc[i]['country_long'].lower():
        count += 1       
print(count)        
```

```python
#which country in african region has highest population
africa_data=data[(data['continent'] == 'Africa')]
africa_data[africa_data['population'] == africa_data['population'].max()]['country']    
```
