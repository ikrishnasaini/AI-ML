```python
from hmmlearn import hmm
import numpy as np
```

```python
states = ['Rainy','Cloudy','Sunny']
n_states = len(states)
observations = ['Umbrella','No Umbrella']
n_observations = len(observations)
```

```python
states_probability = np.array([0.33,0.33,0.34])
transtion_probability = np.array([[0.7,0.3,0.0],
                                  [0.4,0.6,0.0],
                                  [0.0,0.4,0.6]])
emission_probability = np.array([[0.5,0.5],
                                [0.9,0.1],
                                [0.1,0.9]])
```

```python
# Create hmm object
model = hmm.MultinomialHMM(n_components=n_states,n_iter=100)
model.n_trials = 10
model.startprob_ = states_probability
model.transmat_ =transtion_probability
model.emissionprob_ = emission_probability
```

```python
x = np.array([[0,0],[1,1],[1,0],[0,1]])
```

```python
hidden_states = model.predict(x)
```

```python
for s in hidden_states:
    print(states[s])
```
