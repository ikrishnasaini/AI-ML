```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([22,26,28,34,39,42,46])
y = np.array([36000,38000,40000,48000,52000,60000,45000])
```

```python
# Regression line
m, b = np.polyfit(x, y, 1)
y_pred = m * x + b

plt.scatter(x, y, label="Actual Data")
plt.plot(x, y_pred, label="Best Fit Line")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Linear Regression Example")
plt.legend()
plt.show()
```
