```python

def predict_player(time_spent, levels_completed, attempts):
    if time_spent >= 25 and levels_completed >= 3 and attempts >= 5:
        return "Continue"
    else:
        return "Quit"
```

```python
time = int(input("Enter Time Spent: "))
levels_completed = int(input("Enter Levels Completed: "))
attempts = int(input("Enter Attempts: "))
user_prediction = predict_player(time, levels_completed, attempts)
print("Prediction for User Input:", user_prediction)
```

```python
feedback = input("Is this prediction correct? (true/false): ").lower()
```

```python
if feedback == "true":
    new_data = {
        'TimeSpent': [time],
        'LevelsCompleted': [levels_completed],
        'Attempts': [attempts],
        'Result': [user_prediction]
    }

    new_df = pd.DataFrame(new_data)

    # Add new data into old dataset
    _df = pd.concat([df, new_df], ignore_index=True)

    print("\nNew data added successfully!")
else:
    actual_result = input("Please enter the actual result (Continue/Quit): ").capitalize()

    if actual_result in ["Continue", "Quit"]:
        new_data = {
            'TimeSpent': [time],
            'LevelsCompleted': [levels_completed],
            'Attempts': [attempts],
            'Result': [actual_result]
        }

        new_df = pd.DataFrame(new_data)

        # Add new data into old dataset
        _df = pd.concat([df, new_df], ignore_index=True)

        print("\nNew data added successfully!")

        print("\nUpdated Dataset:\n")
        print(_df)

    else:
        print("Invalid Result Entered")



```
