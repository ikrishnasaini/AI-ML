```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Matplotlib Basic Plot
x = np.linspace(1,100,10)
y = x**2
```

```python
plt.plot(x,y)
plt.title("Basic Matplotlib Plot")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

```python
# Subplots in matplotlib
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x**2,y*2)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
```

```python
# Creating a object
```

```python
fig = plt.figure(figsize=(8,8),dpi=100)
axis = fig.add_axes([1,1,.5,.5])
axis.plot(x,y)
axis2 = fig.add_axes([1.7,1,.5,.5])
axis2.plot(y,x)
```

```python
data =pd.read_csv("ipl.csv")
data
```

```python
data.info
```

```python
#Check the size of rows and columns of the dataset
print(f'Number of rows are {data.shape[0]} and Number of columns are {data.shape[1]}')
```

```python
#Now let's see how many columns have null values in total
data.isna().sum()
```

```python
#which team won the most matchs
wins = data['match_winner'].value_counts()
```

```python
sns.barplot(y = wins.index , x=wins.values ,palette='viridis')
```

```python
# toss decision trends
sns.countplot(x =data['toss_decision'],palette='viridis')
```

```python
# Toss winner vs match winner
count = data[data['toss_winner'] == data['match_winner']]['match_id'].count()
percentage = (count*100)/data.shape[0]
percentage.round(2)
```

```python
#How do teams win? (Runs vs Wickets)
sns.countplot( x =data['won_by']  , palette='viridis')
```

```python
w= data[data['won_by'] == 'Wickets'] ['match_id'].count()
w
```

```python
r= data[data['won_by'] == 'Runs'] ['match_id'].count()
r
```

```python
# Most "Player of the Match" Awards
player_of_the_match =data['player_of_the_match'].value_counts()
top_5_player = player_of_the_match.head(5)
sns.barplot(x =top_5_player.values,y = top_5_player.index,palette='viridis')
```

```python
# Most Matches Played by Venue
venue = data['venue'].value_counts()
venue
```

```python
sns.barplot(x =venue.values,y = venue.index,palette='viridis')
```
