```python
raw_prices = ['$15.99','$1,250','$4.50','$100.0']
```

```python
def removal(x):
    x=x.replace('$','')
    x=x.replace(',','')
    x=float(x)
    return x
```

```python
new_price_list = []
for ele in raw_prices:
    new_price_list.append(removal(ele))
```
